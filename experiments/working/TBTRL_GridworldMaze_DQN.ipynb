# TBTRL GridworldMaze DQN

Supported experiment: scratch, sequential TBTRL, all-task evaluation, and independent recovery.

Install from the repository root with `uv sync --extra notebooks --group dev`, then select its Python kernel. The notebook starts with a **smoke run**; set `SMOKE = False` for the configured research budget. Losses and models live in the installed `tbtrl` package.


```python
import os
from pathlib import Path
from uuid import uuid4

from tbtrl.experiments.config import load_config
from tbtrl.experiments.plots import plot_run
from tbtrl.experiments.runner import run_experiment

root = next(
    path
    for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "configs" / "tbtrl_gridworldmaze_dqn.json").is_file()
)
config = load_config(root / "configs" / "tbtrl_gridworldmaze_dqn.json")
SMOKE = True
DEVICE = "cpu"
if SMOKE:
    config = config.smoke()
output_root = Path(os.environ.get("TBTRL_OUTPUT_ROOT", str(root / "runs")))
output = output_root / f"{config.name}-{uuid4().hex[:8]}"
```

```python
records = run_experiment(config, output, device=DEVICE)
print(f"Saved metrics and checkpoints to {output}")
```

```python
figure = plot_run(output)
figure
```

The crosses mark runs that exhausted their budget without meeting the success criterion. The matrix evaluates every goal after each sequential task, including future goals. Smoke-run scores verify execution only; use full budgets, multiple seeds and task orders for scientific conclusions.

Each task directory contains `metrics.json`, sampled `losses.jsonl`, and an evaluation checkpoint. The run manifest records the configuration, seeds, device, package versions, and Git revision. See `docs/method.md` for precisely what is transferred and reset.
